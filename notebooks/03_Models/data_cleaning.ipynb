```python
import pandas as pd

path_master_table = 'data/processed/master_table.csv'
df_master_table = pd.read_csv(path_master_table)
```

```python
import pandas as pd

# Standardize the 'estacion' column
df_master_table['estacion'] = df_master_table['estacion'].str.lower().str.replace(' ', '')

# Group by 'estacion' and calculate the sum of 'pm10' and 'pm2.5'
pm10_sum = df_master_table.groupby('estacion')['pm10'].sum()
pm25_sum = df_master_table.groupby('estacion')['pm2.5'].sum()

# Find the estacion with the highest sum of 'pm10' and 'pm2.5'
max_pm10_estacion = pm10_sum.idxmax()
max_pm25_estacion = pm25_sum.idxmax()

# Print the results
print(f"Estacion with highest pm10 concentration: {max_pm10_estacion}")
print(f"Estacion with highest pm2.5 concentration: {max_pm25_estacion}")

# Create a new CSV with only the information of the estacion with the highest count of concentration
max_estacion = max_pm10_estacion if pm10_sum.max() > pm25_sum.max() else max_pm25_estacion
max_estacion_df = df_master_table[df_master_table['estacion'] == max_estacion]
max_estacion_df.to_csv(f"estacion_data.csv", index=False)

pm10_sum = df_master_table.groupby('estacion')['pm10'].sum()
pm25_sum = df_master_table.groupby('estacion')['pm2.5'].sum()

# Create a new DataFrame with the concentrations of each estacion
concentrations_df = pd.DataFrame({'Estacion': pm10_sum.index, 'PM10': pm10_sum.values, 'PM2.5': pm25_sum.values})

concentrations_df = concentrations_df.sort_values(by=['PM10', 'PM2.5'], ascending=False)

# Print the concentrations of each estacion
print(concentrations_df)
```
